## ArcGIS Intro

Using the data we collected and processed, we can use ArcGIS to produce a nice map like the following
<img src='../images/3_nj_data_center.png' alt='NJ Data Center' width=600/>

ArcGIS is one the many tools for geographic information system (GIS) platforms. Rutgers provides free access to ArcGIS Pro (desktop version) and ArcGIS Online (cloud-based), but there are free software such as [QGIS](https://qgis.org/) as well.

We will use ArcGIS Online to avoid software installation, you can access the portal [here](https://rutgers.maps.arcgis.com).

Esri (originally Environmental Systems Research Institute) provides very great documentation/tutorials, check them out on the [website](https://doc.arcgis.com/en/).

Key steps to illustrate:
1. Navigation
1. Choosing basemap (imagery hybrid)
1. Searching for a location (Piscataway, NJ)
1. Sketching a location (Weeks Hall)
1. Creating a bookmark
1. Adding a layer (Global Land Cover 1992-2020)
    1. Adjust transparency
    1. Adding a filter (before/after a certain year)
    1. Adjusting layer style
1. Save the map
1. Creating an app (lake shrinkage)
    1. Creating a new map (Save and open)
    1. Downloading layers from the [World Imagery Wayback app](https://livingatlas.arcgis.com/wayback/)
    1. See a very nice [example](https://learn.arcgis.com/en/projects/reveal-lake-shrinkage-due-to-severe-drought/) using the the StoryMaps app
