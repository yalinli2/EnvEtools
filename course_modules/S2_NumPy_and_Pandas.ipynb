NumPy and pandas are the two most common Python tools for data analysis:

- **NumPy** for numerical arrays and fast computations
- **pandas** for tabular data analysis

In scientific and engineering work:

- **NumPy** is useful when working with vectors, matrices, and numerical simulation
- **pandas** is useful when working with tables, spreadsheets, time series, and mixed data types

A common workflow is:

1. use **NumPy** for numerical calculations
2. use **pandas** to organize inputs and outputs in table form

```python
import numpy as np
import pandas as pd
```

## NumPy

### Basics

```python
# A Python list
py_list = [1, 2, 3, 4, 5]
print("Python list:", py_list)

# A NumPy array
arr = np.array([1, 2, 3, 4, 5])
print("NumPy array:", arr)
print("Type:", type(arr))
```

NumPy arrays are similar to lists, but they are designed for efficient numerical operations.

```python
# Array properties
arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)
print("Number of dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Data type:", arr.dtype)
print("Size:", arr.size)
```

```python
# 2D arrays
A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(A)
print("Shape:", A.shape)
print("Rows, columns:", A.shape[0], A.shape[1])
```

### Creating arrays

```python
zeros = np.zeros((2, 3))
ones = np.ones((3, 2))
sequence = np.arange(0, 10, 2)
spaced = np.linspace(0, 1, 5)

print("zeros:\n", zeros)
print("\nones:\n", ones)
print("\nsequence:", sequence)
print("spaced:", spaced)
```

`np.arange(start, stop, step)` creates evenly spaced values using a step size.

`np.linspace(start, stop, n)` creates a specified number of evenly spaced values.

### Array math

```python
x = np.array([1, 2, 3, 4])
y = np.array([10, 20, 30, 40])

print("x + y =", x + y)
print("x - y =", x - y)
print("x * y =", x * y)
print("x / y =", x / y)
print("x ** 2 =", x ** 2)
```

```python
# Scalar operations
print("x + 5 =", x + 5)
print("x * 3 =", x * 3)
```

NumPy performs these operations **elementwise**.

### Indexing and slicing

```python
arr = np.array([5, 10, 15, 20, 25, 30])

print("First element:", arr[0])
print("Last element:", arr[-1])
print("Slice arr[1:4]:", arr[1:4])
print("Every other element:", arr[::2])
```

```python
A = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(A)
print("Element at row 1, col 2:", A[1, 2])
print("First row:", A[0, :])
print("Second column:", A[:, 1])
```

### Boolean filtering

```python
arr = np.array([3, 7, 12, 18, 21, 25])

mask = arr > 15
print("Mask:", mask)
print("Values > 15:", arr[mask])
```

Boolean filtering is very useful in data cleaning and selection.

### Basic statistics

```python
data = np.array([12, 15, 14, 10, 18, 20, 16])

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Min:", np.min(data))
print("Max:", np.max(data))
print("Standard deviation:", np.std(data))
print("Sum:", np.sum(data))
```

```python
# Example: daily flow rates in m3/day
flow = np.array([120, 135, 128, 140, 150, 145, 138])

print("Average flow:", flow.mean())
print("Peak flow:", flow.max())
print("Days above 140:", flow[flow > 140])
```

## pandas

### Basics

```python
s = pd.Series([10, 20, 30, 40], name="value")
print(s)
```

A **Series** is a one-dimensional labeled object.

A **DataFrame** is a two-dimensional labeled table.

```python
df = pd.DataFrame({
    "site": ["A", "B", "C", "D"],
    "flow_m3_day": [120, 150, 90, 200],
    "ph": [7.1, 6.8, 7.4, 7.0]
})

df
```

### Inspecting a DataFrame

```python
print(df.head())
print()
print(df.tail())
print()
print(df.info())
print()
print(df.describe())
```

```python
print("Column names:", df.columns.tolist())
print("Shape:", df.shape)
```

### Selecting columns and rows

```python
print(df["site"])
print()
print(df[["site", "flow_m3_day"]])
```

```python
print(df.loc[0])      # label-based row selection
print()
print(df.iloc[1:3])   # position-based row selection
```

### Filtering data

```python
high_flow = df[df["flow_m3_day"] > 120]
high_flow
```

```python
neutral_or_higher = df[df["ph"] >= 7.0]
neutral_or_higher
```

### Adding new columns

```python
df["flow_L_day"] = df["flow_m3_day"] * 1000
df["flow_category"] = np.where(df["flow_m3_day"] >= 130, "high", "low")
df
```

### Sorting

```python
df.sort_values("flow_m3_day", ascending=False)
```

### Grouping and aggregation

```python
df2 = pd.DataFrame({
    "plant": ["North", "North", "South", "South", "South"],
    "month": ["Jan", "Feb", "Jan", "Feb", "Mar"],
    "flow_m3_day": [100, 120, 80, 95, 110]
})

df2
```

```python
df2.groupby("plant")["flow_m3_day"].mean()
```

```python
df2.groupby("plant").agg({
    "flow_m3_day": ["mean", "max", "min"]
})
```

### Missing values

```python
df_missing = pd.DataFrame({
    "site": ["A", "B", "C", "D"],
    "bod_mg_L": [210, np.nan, 180, 195],
    "tss_mg_L": [120, 115, np.nan, 130]
})

df_missing
```

```python
print(df_missing.isna().sum())
print()
print(df_missing.dropna())
print()
print(df_missing.fillna(df_missing.mean(numeric_only=True)))
```

### Reading and writing files

```python
# Save a CSV file
df.to_csv("example_sites.csv", index=False)

# Read it back
df_loaded = pd.read_csv("example_sites.csv")
df_loaded
```

### Using NumPy and pandas together

```python
np.random.seed(42)

measurements = np.random.normal(loc=50, scale=5, size=20)

data_df = pd.DataFrame({
    "sample_id": range(1, 21),
    "measurement": measurements
})

data_df["above_mean"] = data_df["measurement"] > data_df["measurement"].mean()
data_df.head()
```

```python
print("Mean measurement:", data_df["measurement"].mean())
print("Standard deviation:", data_df["measurement"].std())
```
