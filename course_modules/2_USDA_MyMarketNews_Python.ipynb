```python
# Now sorting works
df_to_plot = df_to_plot.sort_values(by='date')
df_to_plot
```

### Plot Price by Type

Now we move to another challenge - creating the grouped column figure as below to see the average prices for different types of cheeses.

Codes below were based on `matplotlib` [grouped bar chart with labels](https://matplotlib.org/stable/gallery/lines_bars_and_markers/barchart.html) example.

<img src="../images/2_price_by_type.png" alt="price by type" width=600/>

```python
# To plot the figure, we need to average the min/max cheese price by type
blue = df_all[df_all.group=='Blue']
blue.price_min.mean() # average price for all blue cheese
```

```python
# Because we need to do this calculation for all types of cheese, we'd better use a loop to repeat:
# First we want to create containers to store the data
min_prices = [] # square brackets indicate a `list` in Python
max_prices = []
cheese_types = df_all.group.unique()
print(cheese_types)
```

```python
# This `nan` means "not a number", this is what `pandas` uses to fill a missing value, we don't like it, so we need to remove it
cheese_types = cheese_types[:-1] # this means we remove the last data from the list
print(cheese_types)
```

```python
# Let's do the loop!
for cheese in cheese_types:
    min_avg = df_all[df_all.group==cheese].price_min.mean()
    max_avg = df_all[df_all.group==cheese].price_max.mean()
    min_prices.append(min_avg) # add the data to list
    max_prices.append(max_avg)
print(f'Min prices are: \n{min_prices}')
print(f'Min prices are: \n{max_prices}')
```

```python
# `numpy` is a power package for array (matrix) computation
import numpy as np
print(f'We have {len(cheese_types)} types of cheese')
x = np.arange(len(cheese_types))  #  this is to create the label locations
print(x)
```

```python
width = 0.25  # the width of the bars
multiplier = 0

# Plot min price first
barfig, barax = plt.subplots()
offset = width * multiplier # how much we need to move for each column
bar_container = barax.bar(x + offset, min_prices, width, label='min price')
# barax.bar_label(bar_container, padding=3) # if want to add data label
```

```python
# Then add the max price
multiplier += 1 # this is the same as multiplier = multiplier + 1
offset = width * multiplier # need to offset for the second set of columns
bar_container = barax.bar(x + offset, max_prices, width, label='max price')
# ax.bar_label(bar_container, padding=3)
barfig
```

```python
# Make it look nicer
barax.set_ylabel('Price [$/lb]')
barax.set_title('Cheese price by type')
barax.set_xticks(x + width/2, cheese_types)
barax.legend(loc='upper left', ncols=2)
barax.set_ylim(0, 3.5)
barax.yaxis.grid(True)
for label in barax.get_xticklabels(which='major'): # rotate labels
    label.set(rotation=45, horizontalalignment='right')
barfig
```

### Plot Price Statistics

To better statistically describe the price data, we can use histrogram and boxplot.

For more help, see [`matplotlib.pyplot.boxplot`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.boxplot.html) and [`matplotlib.pyplot.hist`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.hist.html) docs.

```python
# Let's try box plot first
# Sorting the price by cheese type to prepare the data
# Using the min price as an example
cheese_min_dct = {} # curly brackets indicate dictionary, a type of Python container that contains "key" and "value"
cheese_max_dct = {}

for cheese in cheese_types:
    cheese_min_dct[cheese] = df_all[df_all.group==cheese].price_min
    cheese_max_dct[cheese] = df_all[df_all.group==cheese].price_max
```

```python
cheese_min_dct['Blue'] # you can find the value corresponding to the key
```

```python
# All key-value pairs
cheese_min_dct.items()
```

```python
# Alternatively
for key, value in cheese_min_dct.items(): # sometimes you'll see examples as for k, v
    print(key, value)
    print('---') # to separate key-value pairs
```

```python
# Now we can make the plots
# Since we have two sets of boxes (min and max prices), we need to set the positions of their x values.
offset = 0.3
pos = np.arange(len(cheese_types))*2 + 1
pos1 = pos - offset
pos2 = pos + offset
print(pos1)
print(pos2)
```

```python
# Make plots
bpfig1, bpax1 = plt.subplots()
bplotmin1 = bpax1.boxplot(
    cheese_min_dct.values(),
    positions=pos1,
    widths=0.5,
    patch_artist=True,
    label='min price',
)
bplotmax1 = bpax1.boxplot(
    cheese_max_dct.values(),
    positions=pos2,
    widths=0.5,
    patch_artist=True,
    label='max price',
)
```

```python
# Set labels, let's make a function
def set_labels(bpax):
    for label in bpax.get_xticklabels(which='major'): # rotate labels
        label.set(rotation=45, horizontalalignment='right')
    bpax.set_xlim(0, len(cheese_types)*2)
    bpax.set_ylim(0, 5)
    bpax.set_xticks(pos)
    bpax.set_xticklabels(cheese_min_dct.keys())
    return bpax
bpax1 = set_labels(bpax1)
```

```python
# Now we have a rough draft of the figure, but for some reason the cheddar data is missing
bpfig1
```

```python
# This is because there is empty value (`nan`) in the cheddar data
cheddar_data = cheese_min_dct['Cheddar']
np.isnan(cheddar_data).any()
```

```python
blue_data = cheese_min_dct['Blue']
np.isnan(blue_data).any()
```

```python
# We can use the `dropna` function to throw away the empty data
cheese_min_dct['Cheddar'] = cheese_min_dct['Cheddar'].dropna() # replace the origina data
cheese_max_dct['Cheddar'] = cheese_max_dct['Cheddar'].dropna()
```

```python
np.isnan(cheese_min_dct['Cheddar']).any()
```

```python
# Now we can replot
bpfig2, bpax2 = plt.subplots()
bplotmin2 = bpax2.boxplot(
    cheese_min_dct.values(),
    positions=pos1,
    widths=0.5,
    patch_artist=True,
    label='min price',
)
bplotmax2 = bpax2.boxplot(
    cheese_max_dct.values(),
    positions=pos2,
    widths=0.5,
    patch_artist=True,
    label='max price',
)
set_labels(bpax2)
```

```python
# You'll notice that there are zero values in the cheddar price, you can remove them using the filter function
cheese_min_dct['Cheddar'] = cheese_min_dct['Cheddar'][cheese_min_dct['Cheddar']!=0] # != is not equal to
cheese_max_dct['Cheddar'] = cheese_max_dct['Cheddar'][cheese_max_dct['Cheddar']!=0]
```

```python
# Now the plots don't have zero values
bpfig2, bpax2 = plt.subplots()
bplotmin2 = bpax2.boxplot(
    cheese_min_dct.values(),
    positions=pos1,
    widths=0.5,
    patch_artist=True,
    label='min price',
)
bplotmax2 = bpax2.boxplot(
    cheese_max_dct.values(),
    positions=pos2,
    widths=0.5,
    patch_artist=True,
    label='max price',
)
set_labels(bpax2)
```

```python
# Change box filled color:
for minbox in bplotmin2['boxes']:
    minbox.set_facecolor('peachpuff')
for maxbox in bplotmax2['boxes']:
    maxbox.set_facecolor('tomato')

# Add legend
bpfig2.legend()
bpfig2
```

```python
# We can see that while some prices (e.g., blue max) don't have a lot of "outliers", many of the data do have.
# Histgrams would be another nice way to look at these data

# Let's first look at the blue cheese
blhistfig, blhistax = plt.subplots()
N, bins, patches = blhistax.hist(cheese_min_dct['Blue'])
```

```python
# To make it easier for comparison across different types of cheeses, we'd like to do two things:
# 1. Normalize the count to the total count.
# 2. Adjust the range and bin size.
blhistfig, blhistax = plt.subplots()
N, bins, patches = blhistax.hist(
    cheese_min_dct['Blue'],
    range=(0, 5),
    bins=50, # this is the number of bins
    density=True, # this normalize the count for each bin by the total number of data
    cumulative=False,
)
```

```python
# Now we'd like to plot both min and max for all cheese types
histfig, histaxs = plt.subplots(2, 8, sharex=True, sharey=True) # 2 rows, 8 columns, share x and y axes
for row_num in range(2): # row 0 = min price; row 1 = max price
    if row_num == 0:
        min_or_max = cheese_min_dct
    else:
        min_or_max = cheese_max_dct
    for column_num in range (8): # from blue to swiss cheese
        ax = histaxs[row_num][column_num]
        cheese_data = list(min_or_max.values())[column_num] # the `list` function turns the dict values into a list
        ax.hist(cheese_data, bins=50, range=(0, 5), density=True)
```

```python
# Set x-axis labels
for i in range(len(cheese_types)):
    ax = histaxs[0][i]
    ax.set_title(cheese_types[i], rotation=75)

# Set y-axis labels
histaxs[0][0].set_ylabel('Min Price [$/lb]')
histaxs[1][0].set_ylabel('Max Price [$/kg]')

histfig
```
