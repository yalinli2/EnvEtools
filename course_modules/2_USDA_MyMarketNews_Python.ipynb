```python
# So we can extract the data the same way, instead of copying and pasting the codes, we can create a function to save some efforts
def extract_data(df): # when you starting with `def`, Python knows you are creating a function
    df_new = pd.DataFrame()
    df_new['region'] = df['Column1.region'] # df is the input you provide when using the function
    df_new['date'] = df['Column1.report_date']
    df_new['group'] = df['Column1.group']
    df_new['package'] = df['Column1.package']
    df_new['price_min'] = df['Column1.price_min']
    df_new['price_max'] = df['Column1.price_max']
    return df_new
```

```python
df1083_new = extract_data(df1083)
df1084_new = extract_data(df1084)
df1085_new = extract_data(df1085)
```
