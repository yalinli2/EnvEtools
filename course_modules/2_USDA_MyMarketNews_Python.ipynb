```python
# The better way is to use `matplotlib`, a dedicated tool for plotting
# First to create a new variable for easy referring
df_to_plot = df_all[(df_all['region']=='West') & (df_all['group']=='Process American')]
```

```python
df_to_plot
```

```python
import matplotlib.pyplot as plt # import the `pyplot` module of `matplotlib`
fig, ax = plt.subplots() # create figure and axis objects, you can treat them as containers for the figure and data
# you can't read the tick labels, because they are not recognized as date,
# so pandas tried to label them all, and it takes a lot of time to generate the figure
ax.plot(df_to_plot['date'], df_to_plot['price_min'])
```

```python
df_to_plot.date # `dtype`, data type, is "object"
```

```python
# PS, sorting also doesn't work because the format is wrong
df_to_plot.sort_values(by='date')
```

```python
# So we need to change the data type
df_to_plot.iloc[:, 1] = pd.to_datetime(df_to_plot['date'])
```

```python
# Now sorting works
df_to_plot = df_to_plot.sort_values(by='date')
df_to_plot
```

```python
# We need to make new containers for the figure, otherwise it will be drawn on the previous figure
fig, ax = plt.subplots()
ax.plot(df_to_plot['date'], df_to_plot['price_min'], label='min price')
```

```python
# Make it look nicer
import matplotlib.dates as mdates # this is the same as `from matplotlib import daes as mdates`
ax.xaxis.set_major_locator(mdates.YearLocator(base=2))
ax.xaxis.set_minor_locator(mdates.YearLocator(base=1))
ax.grid(True)
ax.set_ylabel('Price [$/lb]')
ax.set_ylim(0, 3)
for label in ax.get_xticklabels(which='major'): # rotate labels
    label.set(rotation=45, horizontalalignment='right')
fig
```

```python
# Adding the max price is easy enough
ax.plot(df_to_plot['date'], df_to_plot['price_max'], label='max price')
ax.legend()
fig # note that the figure looks different from the one ver
```

### Plot Price by Type

Now we move to another challenge - creating the grouped column figure as below to see the average prices for different types of cheeses.

<img src="../images/2_price_by_type.png" alt="price by type"/>

```python
# To plot the figure, we need to average the min/max cheese price by type
blue = df_all[df_all.group=='Blue']
blue.price_min.mean() # average price for all blue cheese
```

```python
# Because we need to do this calculation for all types of cheese, we'd better use a loop to repeat:
# First we want to create containers to store the data
min_prices = [] # square brackets indicate a `list` in Python
max_prices = []
cheese_types = df_all.group.unique()
print(cheese_types)
```

```python
# This `nan` means "not a number", this is what `pandas` uses to fill a missing value, we don't like it, so we need to remove it
cheese_types = cheese_types[:-1] # this means we remove the last data from the list
print(cheese_types)
```

```python
# Let's do the loop!
for cheese in cheese_types:
    min_avg = df_all[df_all.group==cheese].price_min.mean()
    max_avg = df_all[df_all.group==cheese].price_max.mean()
    min_prices.append(min_avg) # add the data to list
    max_prices.append(max_avg)
print(f'Min prices are: \n{min_prices}')
print(f'Min prices are: \n{max_prices}')
```

```python
# Let's plot!
# https://matplotlib.org/stable/gallery/lines_bars_and_markers/barchart.html
fig, ax = plt.subplots()
```

```python
# `numpy` is a power package for array (matrix) computation
import numpy as np
print(f'We have {len(cheese_types)} types of cheese')
x = np.arange(len(cheese_types))  #  this is to create the label locations
print(x)
```

```python
width = 0.25  # the width of the bars
multiplier = 0

# Plot min price first
offset = width * multiplier # how much we need to move for each column
bar_container = ax.bar(x + offset, min_prices, width, label='min price')
# ax.bar_label(bar_container, padding=3) # if want to add data label
multiplier += 1 # this is the same as multiplier = multiplier + 1

offset = width * multiplier
bar_container = ax.bar(x + offset, max_prices, width, label='max price')
# ax.bar_label(bar_container, padding=3)

ax.set_ylabel('Price [$/lb]')
ax.set_title('Cheese price by type')
ax.set_xticks(x + width, cheese_types)
ax.legend(loc='upper left', ncols=2)
ax.set_ylim(0, 3.5)
ax.yaxis.grid(True)
for label in ax.get_xticklabels(which='major'): # rotate labels
    label.set(rotation=45, horizontalalignment='right')
fig
```
