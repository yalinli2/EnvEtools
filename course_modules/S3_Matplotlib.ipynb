- line plots
- scatter plots
- bar charts
- histograms
- box plots
- plot labels, legends, and figure sizing

`Matplotlib` is useful when you want to:

- visualize trends in data
- compare groups
- check distributions

## Examples

### Basic plotting workflow

A common pattern is:

1. create data
2. create a figure and axes
3. call a plotting function
4. add labels and a title
5. show or save the figure

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 1, 4, 9, 16, 25])

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Basic line plot")
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.show()
```

### Line plots
Line plots are useful for showing changes over time or across an ordered variable.

```python
days = np.array([1, 2, 3, 4, 5, 6, 7])
flow = np.array([120, 135, 128, 140, 150, 145, 138])

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(days, flow, marker="o")
ax.set_title("Daily flow rate")
ax.set_xlabel("Day")
ax.set_ylabel("Flow (m$^3$/day)")
ax.grid(True)
plt.show()
```

### Plotting multiple lines

```python
days = np.arange(1, 8)
influent = np.array([210, 205, 220, 215, 225, 230, 218])
effluent = np.array([32, 30, 28, 35, 31, 29, 33])

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(days, influent, marker="o", label="Influent BOD")
ax.plot(days, effluent, marker="s", label="Effluent BOD")
ax.set_title("Influent and effluent concentrations")
ax.set_xlabel("Day")
ax.set_ylabel("BOD (mg/L)")
ax.legend()
ax.grid(True)
plt.show()
```

### Scatter plots
Scatter plots are useful for checking relationships between two numerical variables.

```python
np.random.seed(0)
x = np.random.uniform(0, 10, 30)
y = 2.5 * x + np.random.normal(0, 3, 30)

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(x, y)
ax.set_title("Scatter plot")
ax.set_xlabel("Input variable")
ax.set_ylabel("Response variable")
ax.grid(True)
plt.show()
```

### Bar charts
Bar charts are useful for comparing categories.

```python
categories = ["Site A", "Site B", "Site C", "Site D"]
values = [120, 150, 90, 200]

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(categories, values)
ax.set_title("Average flow by site")
ax.set_xlabel("Site")
ax.set_ylabel("Flow (m$^3$/day)")
plt.show()
```

### Histograms
Histograms show the distribution of a numerical variable.

```python
np.random.seed(42)
measurements = np.random.normal(loc=50, scale=5, size=200)

fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(measurements, bins=15)
ax.set_title("Histogram of measurements")
ax.set_xlabel("Measurement")
ax.set_ylabel("Frequency")
plt.show()
```

### Box plots
Box plots are useful for comparing medians, spread, and possible outliers across groups.

```python
group1 = np.random.normal(50, 5, 40)
group2 = np.random.normal(55, 7, 40)
group3 = np.random.normal(48, 4, 40)

fig, ax = plt.subplots(figsize=(7, 4))
ax.boxplot([group1, group2, group3], tick_labels=["A", "B", "C"])
ax.set_title("Box plot by group")
ax.set_xlabel("Group")
ax.set_ylabel("Value")
plt.show()
```

### Figure size and labels
Good labels make plots much easier to understand.

```python
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots(figsize=(9, 3))
ax.plot(x, y)
ax.set_title("Sine wave")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
ax.grid(True)
plt.show()
```

### Using `NumPy`/`pandas` with `Matplotlib`

```python
time_hr = np.array([0, 1, 2, 3, 4, 5, 6])
temperature_C = np.array([22, 28, 35, 41, 46, 50, 53])

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(time_hr, temperature_C, marker="o")
ax.set_title("Reactor temperature over time")
ax.set_xlabel("Time (hr)")
ax.set_ylabel("Temperature (°C)")
ax.grid(True)
plt.show()
```

```python
df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "rainfall_mm": [80, 65, 92, 110, 98]
})

df
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(df["month"], df["rainfall_mm"])
ax.set_title("Monthly rainfall")
ax.set_xlabel("Month")
ax.set_ylabel("Rainfall (mm)")
plt.show()
```

### Saving a figure

```python
x = np.linspace(0, 10, 100)
y = x**2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Figure to save")
ax.set_xlabel("x")
ax.set_ylabel("y")

fig.savefig("example_plot.png", dpi=200, bbox_inches="tight")
plt.show()
```

The file `example_plot.png` is now saved in the current working directory.

## Common tips

- use clear axis labels with units
- choose a plot type that matches the question
- avoid clutter
- use legends only when needed
- make sure the title explains what the viewer is seeing
