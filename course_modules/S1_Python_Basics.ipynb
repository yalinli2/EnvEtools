## Core Concepts

### Python

**Python** is a **programming language**.

It defines the rules (syntax) used to write programs.

Python allows you to perform tasks such as:
- numerical calculations
- data analysis
- automation
- scientific modeling
- machine learning

Python code be written/run in many environments, such as:

- Terminal (bash, command prompt, powershell, etc.)
- Jupyter Notebook
- Google Colab
- Development tools (integrated development environment, IDE) like VS Code or PyCharm

So **Python is the core language**, while the other tools are environments used to run Python.

### Jupyter Notebook

**Jupyter Notebook** is an **interactive environment for running Python code**.

Instead of writing one long program, code is organized into **cells**.

A notebook can contain:
- text explanations
- code
- output (tables, numbers, figures)

This format is especially useful for:
- teaching
- data analysis
- experiments
- research notebooks

Notebook files are saved as `.ipynb`

These files store both **code and formatted text**.

#### Running Python in a Notebook

A Jupyter Notebook contains **cells**:
- **Markdown cells**: text and explanation
- **Code cells**: Python code that you can run

Run a cell using:

**Shift + Enter**


### Google Colab

**Google Colab** (short for *Colaboratory*) is a **cloud version of Jupyter Notebook**.

It runs inside a **web browser** and uses **Google's servers** to execute Python code.

Advantages:

- no installation required
- easy sharing
- integrates with Google Drive
- free access to GPUs and cloud computing resources

In practice, Google Colab notebooks look almost identical to Jupyter notebooks.


### Python Script

A **Python script** is a file that contains Python code saved with the extension `.py`


Instead of running code cell-by-cell (like in a notebook), a script runs **all at once from top to bottom**.

Example script: `analysis.py`

You run a Python script from a terminal or command prompt:

```

## Containers

Python provides several container types for storing collections of data. Here we will compare:
- `list`

### List

A **list** is the most common Python container. It stores an ordered sequence of items and can be changed after creation.


````python
rain = [0, 5.2, 12.1, 3.4]

print(rain)
print("First value:", rain[0])
print("Length:", len(rain))

rain.append(7.8)
print("After append:", rain)
````

**List** is typically used for general-purpose data storage, especially when you are adding/removing elements.


### Tuple

A **tuple** is similar to a list, but it cannot be changed after creation.


````python
coordinates = (40.5, -74.4)

print(coordinates)
print("First value:", coordinates[0])
print("Length:", len(coordinates))
````

**Tuple** is typical uses in fixed settings.


````python
list_point = [40.5, -74.4]

tuple_point = (40.5, -74.4)
````

````python
# This works
print(list_point)
list_point[0] = 41.0 
print(list_point)
````

````python
# This triggers error
print(tuple_point)
# tuple_point[0] = 41.0 
# print(tuple_point)
````

Compared with lists, tuples are slightly:
- faster
- more memory-efficient

### Dictionary

A **dictionary** stores data as **key-value pairs**.  
It is very useful when data has labels.


````python
sample = {
    "site": "River A",
    "pH": 7.2,
    "temperature_C": 20.5
}

print(sample)
print("pH:", sample["pH"])

sample["DO_mgL"] = 8.3
print("Updated dictionary:", sample)
````

Typical **dictionary** uses:
- sample metadata
- labeled engineering data
- lookup tables


### Set

A **set** stores unique values only.  
It is useful when duplicates should be removed or when checking whether something is present.


````python
sites = {"A", "B", "C", "A"}

print(sites)
print("Is 'A' in the set?", "A" in sites)
print("Is 'D' in the set?", "D" in sites)
````

**Set** is typical used for:
- removing duplicate values
- membership testing


### Array (`numpy`)

An **array** is designed for efficient numerical computation.  
As the name indicates, it is defined in the `numpy` package, it is commonly used in scientific computing and engineering.


````python
import numpy as np

flow = np.array([1.8, 2.0, 2.2, 2.4])

print(flow)
print("First value:", flow[0])
print("Mean:", np.mean(flow))
print("Flow times 1000:", flow * 1000)
````

````python
data = np.array([1, 2, 3, 4, 5])

print(data)
print(type(data))
````

````python
# Vectorized operations
data = np.array([1, 2, 3, 4, 5])

print(data * 2)
print(data + 10)
````

````python
# Basic statistics
data = np.array([1, 2, 3, 4, 5])

print("Mean:", np.mean(data))
print("Std:", np.std(data))
````

Typical uses:
- numerical vectors
- matrix operations
- scientific calculations


### DataFrame (`pandas`)

A **DataFrame** is a table-like container with labeled rows and columns.  
It is defined in the `pandas` package and is extremely useful for data analysis.


````python
import pandas as pd

data = {
    "site": ["A", "B", "C"],
    "flow_m3s": [2.3, 1.1, 4.0],
    "nitrate_mgL": [3.2, 5.1, 2.8]
}

df = pd.DataFrame(data)

df
````

````python
print("Column names:", df.columns.tolist())
print("Mean flow:", df["flow_m3s"].mean())

print("\nSelect one column:")
print(df["nitrate_mgL"])

print("\nFilter rows with nitrate > 4:")
print(df[df["nitrate_mgL"] > 4])
````

````python
# Basic exploration
print(df.head())
print(df.describe())
````

````python
# Column operations
df["load"] = df["flow_m3s"] * df["nitrate_mgL"]

df
````

Typical uses:
- spreadsheets in Python
- environmental monitoring data
- tabular datasets with multiple columns

## Functions

A **function** is a reusable block of code that performs a specific task.

Functions help make programs:
- easier to read
- easier to reuse
- easier to organize

Instead of repeating the same code many times, we can define a function once and use it whenever needed.


### Basic Syntax

A Python function is defined using the `def` keyword.

```python
def function_name(parameters):
    # code block
    return result

| Part            | Meaning                           |
| --------------- | --------------------------------- |
| `def`           | keyword used to define a function |
| `function_name` | the name of the function          |
| `parameters`    | inputs to the function            |
| `return`        | value sent back by the function   |

````python
def compute_average(data):
    total = 0
    
    for v in data:
        total += v
    
    return total / len(data)

numbers = [10, 20, 30]

print(compute_average(numbers))
````

### `*args` and `**kwargs` in Python

Sometimes we want a function to accept **a flexible number of inputs**.  
Python provides two tools for this:

- `*args` → multiple **positional arguments**
- `**kwargs` → multiple **keyword arguments**

These allow a function to accept **any number of inputs**.


### `*args`: Multiple Positional Arguments

`*args` allows a function to receive **any number of positional arguments**.

````python
# Example function:
def add_numbers(*args):
    print(args)
````

````python
# Calling the function:
add_numbers(1, 2, 3, 4)
````

````python
def sum_numbers(*args):
    total = 0
    for value in args:
        total += value
    return total

sum_numbers(1, 2, 3, 4, 5)
````

In the examples above, `args` becomes a tuple containing all the arguments.

````python
# `*args` collects positional inputs into a **tuple**.

def show_args(*args):
    print(type(args))
    print(args)

show_args(10, 20, 30)
````

### `**kwargs`: Multiple Keyword Arguments

`**kwargs` allows a function to accept **any number of keyword arguments**.

Keyword arguments have **names**.


````python
# Example
def show_info(**kwargs):
    print(kwargs)

show_info(name="River A", pH=7.2, temperature=20)
````

`kwargs` becomes a **dictionary**.

````python
def show_info(**kwargs):
    print(type(kwargs))
    print(kwargs)

show_info(site="A", flow=2.3)
````

### Using *args and **kwargs together

A function can accept **both positional and keyword arguments**.

````python
def example_function(*args, **kwargs):
    print("Positional arguments:", args)
    print("Keyword arguments:", kwargs)

example_function(1, 2, 3, name="River", pH=7.2)
````

**Important rule**

You MUST have positional arguments first. Because `**kwargs` captures all remaining keyword arguments, if you have positional arguments after keyword arguments, Python cannot determine whether a later parameter is for the positional argument, or the keyword argument.

````python
# The following does not work
# def example_function(**kwargs, *args):
#     print("Positional arguments:", args)
#     print("Keyword arguments:", kwargs)
````

### Default Values in Functions

Functions can define **default values** for parameters.

A **default value** is used when the caller does not provide an argument.

Basic syntax:

```python
def function_name(parameter=default_value):
    code block
    return
```

If the argument is provided, Python uses the given value.
If not, Python uses the **default** value.

````python
# Examples
def greet(name="Guest"):
    print("Hello,", name)
````

````python
greet()
````

````python
greet("Alice")
````

````python
def convert_temperature(temp, unit="C"):
    if unit == "C":
        return temp * 9/5 + 32
    else:
        return temp
````

````python
convert_temperature(25)
````

````python
convert_temperature(25, "F")
````

````python
def describe_sample(site="Unknown", pH=7.0, temperature=20):
    print("Site:", site)
    print("pH:", pH)
    print("Temperature:", temperature)
````

````python
describe_sample()
````

````python
describe_sample(site="River A", pH=7.2)
````

````python
describe_sample(site="River A", temperature=25)
````

## Control Flows

**Control flow** is the order in which a program executes instructions.

By default, Python runs code **from top to bottom**.


````python
a = 5
b = 3
print(a + b)
````

The code above runs in order:

1. assign `a`
2. assign `b`
3. print the result


Control flow statements allow Python to **change the normal order** by:

- making decisions
- repeating code
- stopping early
- skipping some steps


In other words, control flow determines:
- whether some code runs or not
- whether code repeats
- when a program stops or skips parts

### Decisions (conditional statements): `if`, `elif`, `else`

Conditional statements let the program make decisions.


````python
pH = 5.8

if pH < 6.5:
    print("Acidic")
elif pH <= 8.5:
    print("Near neutral / typical range")
else:
    print("Basic")
````

#### How it works

- `if` checks the first condition
- `elif` checks another condition if the first one is false
- `else` runs if none of the earlier conditions are true


````python
temperature_C = 32

if temperature_C < 0:
    print("Freezing conditions")
elif temperature_C < 25:
    print("Moderate temperature")
else:
    print("Warm conditions")
````

### Repetitions (loops)

#### `for` loop

A `for` loop repeats code for each item in a container.


````python
rain_mm = [0.0, 5.2, 12.1, 3.4]

for value in rain_mm:
    print(value)
````

This loop runs once for each rainfall value.


````python
sites = ["A", "B", "C"]

for site in sites:
    print("Monitoring site:", site)
````

````python
# Example: compute total rainfall using a loop
rain_mm = [0.0, 5.2, 12.1, 3.4]

total = 0

for value in rain_mm:
    total = total + value

print("Total rainfall =", total)
````

#### `while` loop

A `while` loop repeats as long as a condition stays true.

````python
i = 0

while i < 5:
    print(i)
    i = i + 1
````

Be careful with `while` loops. If the condition never becomes false, the loop will never end.


````python
count = 3

while count > 0:
    print("Countdown:", count)
    count -= 1

print("Done")
````

### Early exit or skipping

Sometimes you may want a program to stop or skip part of a loop.

#### `break`

`break` stops the loop immediately.

````python
for x in range(10):
    if x == 5:
        break
    print(x)
````

#### `continue`

`continue` skips the current iteration and moves to the next one.


````python
rain_mm = [0.0, 5.2, -999, 3.4, 7.1]

for value in rain_mm:
    if value == -999:
        print("Missing value found, skipping")
        continue
    print("Rainfall:", value)
````

### `pass`

`pass` does nothing. It is often used as a temporary placeholder.


You can use `pass` when you want to leave part of the program unfinished for now.


````python
value = 10

if value > 0:
    pass
else:
    print("This will not run")
````

### `return` (inside functions)

`return` ends a function and optionally sends back a value.


````python
def classify_pH(pH):
    if pH < 6.5:
        return "Acidic"
    elif pH <= 8.5:
        return "Near neutral / typical range"
    else:
        return "Basic"

result = classify_pH(7.2)
print(result)
````

````python
def mean_value(data):
    total = 0
    for value in data:
        total += value
    return total / len(data)

print(mean_value([1, 2, 3, 4]))
````

### Basic error handling: `try` / `except`

This allows a program to handle errors more elegantly.


````python
try:
    x = 10 / 0
except:
    print("You cannot divide by zero.")
````

````python
values = ["3.2", "5.1", "bad_data", "4.8"]

for item in values:
    try:
        number = float(item)
        print("Converted:", number)
    except:
        print("Could not convert:", item)
````

### Summary Table

| Category | Statement | Purpose |
|---|---|---|
| Decision | `if`, `elif`, `else` | Run code conditionally |
| Repetition | `for`, `while` | Repeat code |
| Loop control | `break`, `continue` | Stop or skip inside loops |
| Placeholder | `pass` | Do nothing |
| Function control | `return` | Exit a function |
| Error handling | `try`, `except` | Handle errors |

For beginners, focus on these first:

- `if` → make decisions
- `for` → repeat over items
- `while` → repeat while a condition is true

Then learn:

- `break` / `continue`
- `return`
- `try` / `except`
